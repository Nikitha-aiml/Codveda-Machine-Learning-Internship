# Task 1: Data Preprocessing for Machine Learning

## Objective

The objective of this task is to preprocess a raw customer churn dataset and prepare it for machine learning.

The preprocessing steps include:
- Exploring the dataset
- Checking and handling missing values
- Identifying numerical and categorical features
- Encoding categorical variables
- Standardizing numerical features
- Splitting the dataset into training and testing sets

## 1. Import Required Libraries

We use pandas for data manipulation and scikit-learn for preprocessing and train-test splitting.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

## 2. Load the Dataset

The dataset contains customer information and a `Churn` column indicating whether a customer left the service.

```python
df = pd.read_csv("churn-bigml-80.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
```

```python
df.info()
```

## 3. Explore the Dataset

Basic statistical information and data types are examined to understand the structure of the dataset.

```python
df.describe()
```

## 4. Check for Missing Values

Missing values are checked in every column before preprocessing.

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

### Missing Value Observation

The dataset contains no missing values. Therefore, no imputation or row removal is required at this stage.

## 5. Separate Features and Target

```python
# Separate features and target

X = df.drop('Churn', axis=1)
y = df['Churn']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

## 6. Identify Numerical and Categorical Features

Categorical features need to be encoded into numerical values, while numerical features can be standardized.

```python
categorical_features = [
    'State',
    'Area code',
    'International plan',
    'Voice mail plan'
]

numerical_features = [
    col for col in X.columns
    if col not in categorical_features
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

## 7. Split the Dataset into Training and Testing Sets

The dataset is divided into training and testing sets using an 80:20 ratio.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

## 8. Encode Categorical Features and Standardize Numerical Features

One-hot encoding is used for categorical variables. StandardScaler is used to standardize numerical features so that they have a mean close to 0 and a standard deviation close to 1.

The preprocessing transformations are fitted only on the training data to avoid data leakage.

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
print("Processed training data shape:", X_train_processed.shape)
print("Processed testing data shape:", X_test_processed.shape)
```

## 9. Verify the Preprocessed Data

The processed training and testing datasets are now fully numerical and ready to be used by machine learning models.

```python
print("First 5 rows of processed training data:")
print(X_train_processed[:5])
```

```python
# Verify that numerical features are standardized

processed_feature_names = preprocessor.get_feature_names_out()

processed_df = pd.DataFrame(
    X_train_processed,
    columns=processed_feature_names
)

print("Mean of numerical features after scaling:")
print(processed_df.filter(like='num__').mean().head())

print("\nStandard deviation of numerical features after scaling:")
print(processed_df.filter(like='num__').std().head())
```

```python
# Get processed feature names
processed_feature_names = preprocessor.get_feature_names_out()

# Convert processed arrays to DataFrames
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=processed_feature_names,
    index=X_train.index
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=processed_feature_names,
    index=X_test.index
)

# Add target column
X_train_processed_df['Churn'] = y_train.values
X_test_processed_df['Churn'] = y_test.values

# Save processed datasets
X_train_processed_df.to_csv("churn_train_processed.csv", index=False)
X_test_processed_df.to_csv("churn_test_processed.csv", index=False)

print("Processed datasets saved successfully.")
```

##  Verification of Preprocessing

The numerical features were successfully standardized, with their means close to 0 and standard deviations close to 1.

The categorical features were converted into numerical columns using one-hot encoding.

The training and testing datasets now contain the same 73 processed features and are ready for use in machine learning models.

## Conclusion

The customer churn dataset was successfully preprocessed for machine learning.

The preprocessing included:
- Checking for missing values
- Identifying numerical and categorical features
- Separating features and target
- Splitting the data into training and testing sets
- One-hot encoding categorical features
- Standardizing numerical features
- Preventing data leakage by fitting preprocessing transformations only on the training data

The resulting processed data is ready for machine learning model development.
